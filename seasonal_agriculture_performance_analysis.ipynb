```python
# Data Types Analysis

data.dtypes
print("Numerical Columns:")
print(data.select_dtypes(include=np.number).columns.tolist())

print("\nCategorical Columns:")
print(data.select_dtypes(include="object").columns.tolist())
```

```python
# Missing Value Analysis

# Missing values are checked to determine whether data cleaning is required.

data.isnull().sum()
```

```python
# Data Cleaning & Preprocessing

data['Rainfall_mm'] = data['Rainfall_mm'].fillna(data['Rainfall_mm'].median())
data['Soil_Moisture_pct'] = data['Soil_Moisture_pct'].fillna(data['Soil_Moisture_pct'].median())
data['Yield_Tonnes_Ha'] = data['Yield_Tonnes_Ha'].fillna(data['Yield_Tonnes_Ha'].median())
```

```python
data.isnull().sum()
```

```python
# Duplicate Records

# Duplicate records are identified to ensure the reliability of the analysis.

print("Number of duplicate rows:", data.duplicated().sum())
```

```python
# Data Types

type_table = pd.DataFrame({
    "Column": data.columns,
    "Data_Type": data.dtypes.astype(str),
    "Unique_Values": [data[col].nunique() for col in data.columns]
})

type_table

categorical_columns = data.select_dtypes(include="object").columns.tolist()
numerical_columns = data.select_dtypes(include=np.number).columns.tolist()

print("Categorical variables:")
print(categorical_columns)

print("\nNumerical variables:")
print(numerical_columns)
```

```python
# Statistical Summary

# Descriptive statistics are used to understand the central tendency, spread and range of numerical variables.

statistical_analysis = pd.DataFrame({
    "Mean": data[numerical_columns].mean(),
    "Median": data[numerical_columns].median(),
    "Standard_Deviation": data[numerical_columns].std(),
    "Minimum": data[numerical_columns].min(),
    "Maximum": data[numerical_columns].max()
})

statistical_analysis

data[numerical_columns].describe().T
```

```python
# Outlier Investigation

# Box plots are used to investigate unusually high or low values in important numerical variables.

outlier_columns = [
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Profit_INR",
    "Water_Used_m3"
]

for column in outlier_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=data[column])
    plt.title("Outlier Investigation - " + column)
    plt.xlabel(column)
    plt.tight_layout()
    plt.show()
```

```python
# Unique Values in Categorical Variables

print("States:", data["State"].nunique())
print("Districts:", data["District"].nunique())
print("Crops:", data["Crop"].nunique())
print("Seasons:", data["Season"].nunique())
print("Irrigation Methods:", data["Irrigation_Method"].nunique())
```

```python
# Different Categories in our dataset

print("Crops:")
print(data["Crop"].unique())

print("\nSeasons:")
print(data["Season"].unique())

print("\nIrrigation Methods:")
print(data["Irrigation_Method"].unique())
```

```python
# Overall Agriculture Performance

# Key performance indicators are calculated to obtain an overall understanding of agricultural production, revenue, cost, profit and resource utilization.

print("Total Farms:", data["Farm_ID"].nunique())
print("Total Agricultural Area:", round(data["Farm_Area_Hectares"].sum(), 2), "hectares")
print("Total Production:", round(data["Production_Tonnes"].sum(), 2), "tonnes")
print("Total Revenue: ₹", round(data["Revenue_INR"].sum(), 2))
print("Total Cost: ₹", round(data["Total_Cost_INR"].sum(), 2))
print("Total Profit: ₹", round(data["Profit_INR"].sum(), 2))
print("Average Yield:", round(data["Yield_Tonnes_Ha"].mean(), 2), "tonnes/hectare")
print("Average Disease/Pest Risk:", round(data["Disease_Pest_Risk_pct"].mean(), 2), "%")
print("Total Water Used:", round(data["Water_Used_m3"].sum(), 2), "m³")
```

```python
# Univariate Analysis

# Individual variables are analyzed separately to understand their distributions and frequency patterns.

# Crop distribution

crop_distribution = data["Crop"].value_counts()

print(crop_distribution)
plt.figure(figsize=(10, 6))
plt.bar(crop_distribution.index, crop_distribution.values)
plt.title("Farm Records by Crop")
plt.xlabel("Crop")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Irrigation distribution

irrigation_distribution = data["Irrigation_Method"].value_counts()

print(irrigation_distribution)
plt.figure(figsize=(8, 5))
plt.bar(
    irrigation_distribution.index,
    irrigation_distribution.values
)
plt.title("Distribution of Irrigation Methods")
plt.xlabel("Irrigation Method")
plt.ylabel("Number of Records")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Season distribution

season_counts = data["Season"].value_counts()

print(season_counts)
plt.figure(figsize=(8, 5))
season_counts.plot(kind="bar")
plt.title("Distribution of Agricultural Seasons")
plt.xlabel("Season")
plt.ylabel("Number of Farms")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Yield distribution

plt.figure(figsize=(8, 5))
plt.hist(data["Yield_Tonnes_Ha"], bins=30)
plt.title("Distribution of Crop Yield")
plt.xlabel("Yield (Tonnes/Ha)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Profit distribution

plt.figure(figsize=(9, 5))
plt.hist(data["Profit_INR"], bins=40)
plt.title("Distribution of Farm Profit")
plt.xlabel("Profit (INR)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Bivariate Analysis

# Relationships between pairs of variables are analyzed to understand how one agricultural factor is associated with another.

# Rainfall vs Yield

plt.figure(figsize=(8, 6))
plt.scatter(
    data["Rainfall_mm"],
    data["Yield_Tonnes_Ha"],
    alpha=0.5
)
plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
# Farm area vs production

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=data,
    x="Farm_Area_Hectares",
    y="Production_Tonnes"
)

plt.title("Farm Area vs Total Production")
plt.xlabel("Farm Area (Hectares)")
plt.ylabel("Production (Tonnes)")
plt.tight_layout()
plt.show()
```

```python
# Water Efficiency vs Crop Yield

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=data,
    x="Water_Efficiency_t_per_1000m3",
    y="Yield_Tonnes_Ha",
    alpha=0.5
)

plt.title("Water Efficiency vs Crop Yield")
plt.xlabel("Water Efficiency (Tonnes per 1000 m³)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
# Fertilizer vs Yield

plt.figure(figsize=(8, 6))
plt.scatter(
    data["Fertilizer_kg_ha"],
    data["Yield_Tonnes_Ha"],
    alpha=0.5
)
plt.title("Fertilizer Usage vs Crop Yield")
plt.xlabel("Fertilizer (kg/ha)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
# Seed Quality vs Yield

plt.figure(figsize=(8, 6))
plt.scatter(
    data["Seed_Quality_Score"],
    data["Yield_Tonnes_Ha"],
    alpha=0.5
)
plt.title("Seed Quality vs Crop Yield")
plt.xlabel("Seed Quality Score")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
# Revenue vs Cost

plt.figure(figsize=(8, 6))
plt.scatter(
    data["Total_Cost_INR"],
    data["Revenue_INR"],
    alpha=0.5
)
plt.title("Total Cost vs Revenue")
plt.xlabel("Total Cost (INR)")
plt.ylabel("Revenue (INR)")
plt.tight_layout()
plt.show()
```

```python
# Multivariate Analysis

# Multiple agricultural variables are analyzed together to understand their combined patterns and relationships with crop yield.

multi_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Soil_Moisture_pct",
    "Fertilizer_kg_ha",
    "Yield_Tonnes_Ha"
]

season_multi = data[multi_columns + ["Season"]]

sns.pairplot(
    season_multi.sample(
        min(600, len(season_multi)),
        random_state=10
    ),
    hue="Season"
)

plt.show()
```

```python
# Multivariate crop analysis

season_crop_matrix = pd.pivot_table(
    data,
    values="Yield_Tonnes_Ha",
    index="Crop",
    columns="Season",
    aggfunc="mean"
)

season_crop_matrix
```

```python
# Correlation Analysis

# Correlation analysis is performed to identify the strength and direction of relationships between numerical agricultural variables.

numeric_data = data.select_dtypes(include=np.number)

correlation = numeric_data.corr()

correlation
```

```python
# Heatmap

plt.figure(figsize=(18, 12))
sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Yield correlation

correlation_matrix = data[numerical_columns].corr()

yield_relationships = (
    correlation_matrix["Yield_Tonnes_Ha"]
    .sort_values(ascending=False)
)

yield_relationships
```

```python
# Seasonal Comparison

# Agricultural performance is compared across different seasons using yield, production, revenue, profit, water efficiency and disease or pest risk.

season_analysis = data.groupby("Season").agg(
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Total_Production=("Production_Tonnes", "sum"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Water_Use=("Water_Used_m3", "mean"),
    Average_Water_Efficiency=(
        "Water_Efficiency_t_per_1000m3",
        "mean"
    ),
    Average_Risk=("Disease_Pest_Risk_pct", "mean")
)

season_analysis.round(2)
```

```python
season_analysis[
    ["Average_Yield", "Average_Profit", "Average_Water_Efficiency"]
].plot(
    kind="bar",
    subplots=True,
    layout=(1, 3),
    figsize=(15, 5),
    legend=False
)
plt.tight_layout()
plt.show()
```

```python
# Seasonal yield

plt.figure(figsize=(8, 5))
season_analysis["Average_Yield"].plot(kind="bar")
plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Seasonal profit

plt.figure(figsize=(8, 5))
season_analysis["Average_Profit"].plot(kind="bar")
plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Profit (INR)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Seasonal risk

plt.figure(figsize=(8, 5))
season_analysis["Average_Risk"].plot(kind="bar")
plt.title("Disease and Pest Risk by Season")
plt.xlabel("Season")
plt.ylabel("Risk (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Crop Profitability and Market Performance

crop_economic_analysis = data.groupby("Crop").agg(
    Mean_Market_Price=("Market_Price_INR_Tonne", "mean"),
    Mean_Revenue=("Revenue_INR", "mean"),
    Mean_Cost=("Total_Cost_INR", "mean"),
    Mean_Profit=("Profit_INR", "mean")
).sort_values(
    "Mean_Profit",
    ascending=False
)

crop_economic_analysis.round(2)

plt.figure(figsize=(11, 6))

plt.bar(
    crop_economic_analysis.index,
    crop_economic_analysis["Mean_Profit"]
)

plt.title("Average Profitability by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Irrigation Efficiency and Productivity

irrigation_performance = data.groupby(
    "Irrigation_Method"
).agg(
    Mean_Yield=("Yield_Tonnes_Ha", "mean"),
    Mean_Water_Use=("Water_Used_m3", "mean"),
    Mean_Water_Efficiency=(
        "Water_Efficiency_t_per_1000m3",
        "mean"
    ),
    Mean_Profit=("Profit_INR", "mean")
).sort_values(
    "Mean_Water_Efficiency",
    ascending=False
)

irrigation_performance.round(2)

plt.figure(figsize=(9, 6))

sns.barplot(
    data=irrigation_performance.reset_index(),
    x="Irrigation_Method",
    y="Mean_Water_Efficiency"
)

plt.title("Water Efficiency Across Irrigation Methods")
plt.xlabel("Irrigation Method")
plt.ylabel("Water Efficiency")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Disease Risk and Agricultural Performance

risk_analysis = data.groupby("Season").agg(
    Mean_Risk=("Disease_Pest_Risk_pct", "mean"),
    Mean_Yield=("Yield_Tonnes_Ha", "mean"),
    Mean_Profit=("Profit_INR", "mean")
).sort_values(
    "Mean_Risk",
    ascending=False
)

risk_analysis.round(2)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=risk_analysis.reset_index(),
    x="Season",
    y="Mean_Risk"
)

plt.title("Disease and Pest Risk Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Risk (%)")
plt.tight_layout()
plt.show()
```

```python
# Crop-level risk

crop_risk = data.groupby("Crop")[
    "Disease_Pest_Risk_pct"
].mean().sort_values(ascending=False)

crop_risk
```

```python
# Average Yield by Crop

# The average yield of each crop is compared to identify high-performing crops.

crop_yield = data.groupby("Crop")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)

print(crop_yield)

plt.figure(figsize=(10, 6))
crop_yield.plot(kind="bar")
plt.title("Average Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Average Profit by Crop

# The profitability of different crops is analyzed using average profit per farm.

crop_profit = data.groupby("Crop")["Profit_INR"].mean().sort_values(ascending=False)

print(crop_profit)
```

```python
# Average Profit Graph

plt.figure(figsize=(10, 6))
crop_profit.plot(kind="bar")
plt.title("Average Profit by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# State-wise Agricultural Performance

# Agricultural performance is compared across different states.

state_summary = data.groupby("State").agg(
    Farms=("Farm_ID", "count"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Total_Production=("Production_Tonnes", "sum"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Water_Used=("Water_Used_m3", "mean"),
    Average_Risk=("Disease_Pest_Risk_pct", "mean")
).sort_values("Average_Profit", ascending=False)

state_summary
```

```python
# Top 10 States by Profit

top_states = state_summary.head(10)

print(top_states)
```

```python
# State Profit Graph

plt.figure(figsize=(12, 6))
top_states["Average_Profit"].plot(kind="bar")
plt.title("Top States by Average Profit")
plt.xlabel("State")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Irrigation Method Analysis

# Different irrigation methods are compared based on yield, profit and water efficiency.

irrigation_summary = data.groupby("Irrigation_Method").agg(
    Farms=("Farm_ID", "count"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Water_Used=("Water_Used_m3", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean")
).sort_values("Average_Water_Efficiency", ascending=False)

irrigation_summary
```

```python
# Irrigation Water Efficiency Graph

plt.figure(figsize=(10, 6))
irrigation_summary["Average_Water_Efficiency"].plot(kind="bar")
plt.title("Water Efficiency by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Water Efficiency (Tonnes per 1000 m³)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
best_yield_crop = (
    data.groupby("Crop")["Yield_Tonnes_Ha"]
    .mean()
    .idxmax()
)

best_profit_crop = (
    data.groupby("Crop")["Profit_INR"]
    .mean()
    .idxmax()
)

best_water_method = (
    data.groupby("Irrigation_Method")
    ["Water_Efficiency_t_per_1000m3"]
    .mean()
    .idxmax()
)

best_profit_season = (
    data.groupby("Season")["Profit_INR"]
    .mean()
    .idxmax()
)

highest_risk_season = (
    data.groupby("Season")["Disease_Pest_Risk_pct"]
    .mean()
    .idxmax()
)

print("Highest average yield crop:", best_yield_crop)
print("Highest average profit crop:", best_profit_crop)
print("Most water-efficient irrigation method:", best_water_method)
print("Most profitable season:", best_profit_season)
print("Highest disease/pest risk season:", highest_risk_season)
```
